```python
import splitfolders
```

```python
input_folder = "../dataset/cropped"
output_folder = "../dataset/split"
```

```python
splitfolders.ratio(
    input_folder,
    output=output_folder,
    seed=42,
    ratio=(0.8, 0.2)
)
```

```python
import os

for split_name in ["train", "test"]:
    print(f"\n{split_name.upper()}")

    total = 0

    for class_name in ["glioma", "meningioma", "pituitary"]:
        folder = os.path.join(output_folder, split_name, class_name)

        count = len([
            f for f in os.listdir(folder)
            if f.lower().endswith(".png")
        ])

        print(f"{class_name}: {count}")
        total += count

    print("Total:", total)
```
